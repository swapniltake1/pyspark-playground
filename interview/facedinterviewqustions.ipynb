# Write a SQL query to calculate total purchase amount for each customer from w3 school. 

```python


SELECT 
    c.CustomerID,
    c.CustomerName,
    SUM(od.Quantity * p.Price) AS total_purchase_amount
FROM Customers c
JOIN Orders o 
    ON c.CustomerID = o.CustomerID
JOIN OrderDetails od 
    ON o.OrderID = od.OrderID
JOIN Products p 
    ON od.ProductID = p.ProductID
GROUP BY 
    c.CustomerID, c.CustomerName
ORDER BY 
    total_purchase_amount DESC;
```

# Write a Python function to calculate the final position of a drone in 3D space based on movement instructions with a 1000-unit limit.

```python
def navigate_drone(starting_coordinate, instructions):
    x, y, z = starting_coordinate
    total_movement = 0
    
    for instruction in instructions:
        if total_movement >= 1000:   # ✅ FIX 1: check before processing
            break
        
        distance = int(instruction[:-1])
        direction = instruction[-1]
        
        magnitude = abs(distance)
        allowed = min(magnitude, 1000 - total_movement)   # ✅ FIX 2: enforce limit
        actual_distance = allowed if distance >= 0 else -allowed
        
        # ✅ FIX 3: apply movement using actual_distance
        if direction == 'N':
            y += actual_distance
        elif direction == 'S':
            y -= actual_distance
        elif direction == 'E':
            x += actual_distance
        elif direction == 'W':
            x -= actual_distance
        elif direction == 'U':
            z += actual_distance
        elif direction == 'D':
            z -= actual_distance
        
        total_movement += allowed   # ✅ FIX 4: update correctly
    
    return [x, y, z]


if __name__ == '__main__':
    starting_coordinate = [500, 400, 150]
    instructions = ['4N', '3E', '2U', '4S', '5S', '3E', '1D']
    print(navigate_drone(starting_coordinate, instructions))
```
