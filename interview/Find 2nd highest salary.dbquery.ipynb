```sql
-- Method 1: Using LIMIT (Simple Way)
-- Get distinct salaries in descending order
-- Skip highest using subquery

SELECT salary
FROM (
    SELECT DISTINCT salary
    FROM main.default.employees
    ORDER BY salary DESC
    LIMIT 2
) t
ORDER BY salary
LIMIT 1;


-- Method 2: Using DENSE_RANK()

-- Rank salaries in descending order
-- Pick rank = 2

SELECT salary
FROM (
    SELECT salary,
           DENSE_RANK() OVER (ORDER BY salary DESC) AS rnk
    FROM main.default.employees
) t
WHERE rnk = 2;
```
